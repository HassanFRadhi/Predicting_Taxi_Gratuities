### **Part 2c. Understanding the data - Investigating the variables**

We'll sort and interpret the data table for two key variables: `trip_distance` and `total_amount`.

1. Sorting the first variable (`trip_distance`) from maximum to minimum value, we found that the values are greatly larger than the mean.
2. Sorting by the second variable (`total_amount`), we observed outliers and negative values.
3. Greater distance doesn't seem to reflect a higher total amount in the results.

```python
# Sorting the data by trip distance from maximum to minimum value
df.sort_values('trip_distance', ascending=False).head()
```

```python
# Sorting the data by total amount and printing the top 20 values
df.sort_values('total_amount', ascending=False).head(20)
```

```python
# Sorting the data by total amount and printing the bottom 20 values
df.sort_values('total_amount').head(20)
df[df['total_amount'] < 0].groupby('payment_type').size()
```

```python
# Checking how many of each payment type are represented in the data
df.groupby('payment_type').size()
```

According to the data dictionary, the payment method was encoded as follows:

1 = Credit card  
2 = Cash  
3 = No charge  
4 = Dispute  
5 = Unknown  
6 = Voided trip

```python
# Finding the average tip for trips paid for with a credit card
print(df[df['payment_type'] == 1]['tip_amount'].mean())

# Finding the average tip for trips paid for with cash
df[df['payment_type'] == 2]['tip_amount'].mean()
```

```python
# Checking how many times each vendor ID is represented in the data
df.groupby('VendorID').size()
```

```python
# Calculating the mean total amount for each vendor
df.groupby('VendorID')['total_amount'].mean()
```

```python
# Filtering the data for credit card payments only
cc_df = df[df['payment_type'] == 1]

# Filtering credit-card-only data for passenger count only
pcc_df = cc_df[['passenger_count']]
```

```python
# Calculating the average tip amount for each passenger count (credit card payments only)
cc_df.groupby('passenger_count')['tip_amount'].mean()
```
