### Part 1. Imports and data loading

First, we will import the packages and libraries needed to compute descriptive statistics and conduct our hypothesis test.

```python
# Importing required libraries for statistical analysis and manipulation
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels as st
```

```python
# Loading the dataset into a pandas dataframe
data = pd.read_csv("2017_Yellow_Taxi_Trip_Data.csv", index_col = 0)
```

## PACE: **Analyze and Construct**

Descriptive statistics help us determine the important characeteristics of the data such as the center and dispersion which helps us determine how our data is distributed and centered.

### Part 2. Data exploration

Let's use descriptive statistics to perform our Exploratory Data Analysis (EDA).

In the dataset, `payment_type` is encoded in integers:
*   1: Credit card
*   2: Cash
*   3: No charge
*   4: Dispute
*   5: Unknown



```python
# Converting the payment type column to category and describing it, then gathering value distribution
data['payment_type'].astype('category').describe()
data['payment_type'].value_counts(normalize=True)
```

We are interested in the relationship between how a customer pays and the fare amount they pay. Our chosen approach is to look at the average fare amount for each payment type.

```python
# Grouping by payment type and calculating the mean fare amount
data.groupby('payment_type')['fare_amount'].mean()
```

Based on the averages we computed, it appears that customers who pay with a credit card tend to pay a larger fare amount than customers who pay with cash. However, this difference might arise from random sampling, rather than being a true difference in fare amount. To assess whether the difference is statistically significant, we will conduct a hypothesis test.

### Part 3. Hypothesis testing

Before conducting our hypothesis test, we should define our hypotheses:

$H_0$: There is no difference in the average fare amount between customers who use credit cards and customers who use cash.

$H_A$: There is a difference in the average fare amount between customers who use credit cards and customers who use cash.

Our goal in this step is to run a two-sample t-test. Let's recall the standard procedure:

1. State our null hypothesis and alternative hypothesis.
2. Select our significance level.
3. Compute our p-value.
4. Determine whether to reject or fail to reject the null hypothesis.

For our purposes here, this hypothesis test serves as the main component of our A/B test.

We choose 5% as our significance level and proceed to perform our two-sample t-test.

```python
# Running a two-sample t-test comparing credit card payments vs cash payments
significance_level = 0.05
stats.ttest_ind(a=data[data['payment_type']==1]['fare_amount'], b=data[data['payment_type']==2]['fare_amount'], equal_var=False)
```

> P-value = 6.8 * $10^{-12}$ , which is less than the significance level. Therefore, we `Reject` the null hypothesis
